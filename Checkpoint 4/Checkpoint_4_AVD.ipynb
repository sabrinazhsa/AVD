## **Cek Outliers**

```python
results = []

cols = df.select_dtypes(include=['float64', 'int64'])

for col in cols:
  q1 = df[col].quantile(0.25)
  q3 = df[col].quantile(0.75)
  iqr = q3 - q1
  lower_bound = q1 - 1.5*iqr
  upper_bound = q3 + 1.5*iqr
  outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
  percent_outliers = (len(outliers)/len(df))*100
  results.append({'Kolom': col, 'Persentase Outliers': percent_outliers})

# Dataframe dari list hasil
results_df = pd.DataFrame(results)
results_df.set_index('Kolom', inplace=True)
results_df = results_df.rename_axis(None, axis=0).rename_axis('Kolom', axis=1)

# Tampilkan dataframe
display(results_df)
```

Pada data di atas dapat dilihat bahwa pada kolom age dan rating memiliki nilai 0,0%, artinya seluruh data stabil dan tergolong baik karena tidak ada data anomali. Sedangkan pada kolom years_of_experience terdapat nilai 0,9%, di mana ini menunjukkan adanya outliers. Persentase ini sangat kecil sehingga tidak memengaruhi terhadap keseluruhan analisis data, karena jumlahnya yang masih kurang dari 1%.

1. **Age**

Untuk data kolom age, missing values sebanyak 3% di mana jika missing values >= 70% maka akan dilakukan penanganan berupa dihapus, tetapi jika missing values <= 70% maka akan dilakukan penanganan berupa diimputasi. Pada kasus ini diperlukan penanganan berupa diimputasi menggunakan median (nilai tengah) karena data berbentuk numerik dan memiliki rentang nilai yang luas serta tidak dihapus karena missing values di bawah 70%.

```python
df['age'] = df['age'].fillna(df['age'].dropna().median())
```

2. **years_of_experience**

Untuk data kolom years_of_experience, missing values sebanyak 5%. Pada kasus ini diperlukan penanganan menggunakan median (nilai tengah) karena data berbentuk numerik dan memiliki rentang nilai yang luas serta tidak dihapus karena missing values di bawah 70%.

```python
df['years_of_experience'] = df['years_of_experience'].fillna(df['years_of_experience'].dropna().median())
```

3. **hourly_rate** **(USD)**

Untuk data kolom hourly_rate (USD), missing values sebanyak 9%. Pada kasus ini diperlukan penanganan menggunakan median (nilai tengah) karena data berbentuk numerik dan memiliki rentang nilai yang luas serta tidak dihapus karena missing values di bawah 70%.

```python
df['hourly_rate (USD)'] = pd.to_numeric(df['hourly_rate (USD)'], errors='coerce')
df['hourly_rate (USD)'] = df['hourly_rate (USD)'].fillna(df['hourly_rate (USD)'].dropna().median())
```

4. **rating**

Untuk data kolom rating, missing values sebanyak 10%. Pada kasus ini diperlukan penangnanan menggunakan mean (rata-rata) karena data berbentuk numerik dan rentang nilai yang kecil (skala nilai 1-5) serta tidak dihapus karena missing values di bawah 70%.

```python
df['rating'] = df['rating'].fillna(df['rating'].dropna().mean())
```

5. **is_active**

Untuk data kolom is_active, missing values sebanyak 8%. Pada kasus ini diperlukan penanganan menggunakan modus (nilai yang paling sering muncul) karena data berbentuk kategorikal sehingga mean dan median tidak dapat digunakan, serta tidak dihapus karena missing values di bawah 70%.

```python
df['is_active'] = df['is_active'].fillna(df['is_active'].dropna().mode()[0])
```

6. **client_satisfaction**

Untuk data kolom client_satisfaction, missing values sebanyak 17%. Pada kasus ini diperlukan penanganan menggunakan median (nilai tengah) karena data berbentuk numerik dan memiliki rentang nilai yang luas serta tidak dihapus karena missing values di bawah 70%.

```python
df['client_satisfaction'] = df['client_satisfaction'].fillna(df['client_satisfaction'].dropna().median())
```
